# 1. Exploratory Data Analysis
## 1.1 Univariate Analysis
1.1.1 Rating Distribution


```python
plt.figure(figsize=(5, 4))

sns.histplot(data=df, x="Avg ratings", bins=20, kde=True)

plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Average Rating")
plt.ylabel("Number of Restaurants")

plt.show()
```

### Insight: Distribution of Restaurant Ratings

- Most restaurants have ratings close to 3.0, forming a prominent peak in the distribution.
- Another concentration of restaurants appears between 4.0 and 4.5, indicating many highly rated restaurants.
- Ratings between 3.5 and 4.5 are relatively common, while ratings below 2.5 and above 4.8 are rare.
- The distribution is bimodal, with two distinct concentrations around 3.0 and 4.2.

**Business Implication:** Further analysis of cuisine, pricing, and delivery time may help explain the differences between moderately rated and highly rated restaurants.

1.2 Cuisine popularity

```python
top_food_types = df["Food type"].value_counts().head(10)

plt.figure(figsize=(8, 4))

sns.barplot(x=top_food_types.values, y=top_food_types.index,)

plt.title("Top 10 Most Common Food Types")
plt.xlabel("Number of Restaurants")
plt.ylabel("Food Type")

plt.show()
```

### Insight: Most Common Food Types

- Indian cuisine is the most common food type, with approximately 390 restaurants.
- Chinese cuisine ranks second, followed by North Indian, Fast Food, and South Indian.
- Indian and Chinese cuisine combinations also appear frequently in the dataset.
- Desserts, bakery items, and beverages have comparatively fewer restaurant listings among the top 10 categories.

**Business Implication:** Indian cuisine has a strong presence in the restaurant market. Restaurants offering popular cuisines may face greater competition, while less represented categories could provide opportunities for market differentiation.

1.3 Price distribution

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Price", bins=30, kde=True)

plt.title("Distribution of Restaurant Prices")
plt.xlabel("Price")
plt.ylabel("Number of Restaurants")

plt.show()
```

### Insight: Distribution of Restaurant Prices

- Most restaurants have prices concentrated between ₹200 and ₹400.
- The distribution is right-skewed, indicating that relatively few restaurants have significantly higher prices.
- A small number of restaurants have prices exceeding ₹1,000, representing the premium-priced segment.
- The dataset suggests that affordable and mid-range restaurants dominate the market.

**Business Implication:** Restaurants targeting the ₹200–₹400 price segment may appeal to a larger portion of the market, while premium restaurants can differentiate themselves through specialized offerings and higher-value experiences.

1.4 Delivery time distribution

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Delivery time", bins=20, kde=True)

plt.title("Distribution of Restaurant Delivery Times")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Restaurants")

plt.show()
```

### Insight: Distribution of Restaurant Delivery Times

- Most restaurants have delivery times concentrated between 40 and 60 minutes.
- The highest concentration of restaurants is around 45–55 minutes.
- The distribution is right-skewed, with a smaller number of restaurants taking more than 80 minutes.
- Delivery times range from approximately 20 to 100 minutes, indicating considerable variation in delivery speed.

**Business Implication:** Improving delivery efficiency for restaurants with longer delivery times could enhance customer experience and help reduce customer dissatisfaction.

1.5 restaurant ratings across different price ranges.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Price",
    y="Avg ratings",
    alpha=0.4
)

plt.title("Restaurant Price vs. Average Rating")
plt.xlabel("Price")
plt.ylabel("Average Rating")

plt.show()
```

### Insight: Restaurant Price vs. Average Rating

- Most restaurants are concentrated in the lower-to-mid price range, with ratings generally between 3.5 and 4.8.
- Higher prices do not consistently correspond to higher ratings. Several expensive restaurants have ratings similar to more affordable restaurants.
- Many restaurants across different price ranges have ratings close to 2.9, suggesting a possible rating pattern worth investigating.
- Highly rated restaurants are available at relatively affordable prices, indicating that price alone does not determine customer satisfaction.

**Business Implication:** Restaurants should focus on food quality and customer experience rather than relying solely on higher prices to improve ratings. Further analysis of cuisine and delivery time may help explain rating differences.

## Conclusion

This exploratory data analysis examined 8,680 restaurant listings across nine cities to understand restaurant ratings, cuisine popularity, pricing, and delivery times.

### Key Findings

- **Ratings:** Restaurant ratings show two main concentrations, around 3.0 and 4.0–4.5. Ratings between 3.5 and 4.5 are common.
- **Cuisine popularity:** Indian cuisine is the most frequently listed food type, followed by Chinese and North Indian.
- **Pricing:** Most restaurants fall within the ₹200–₹400 price range. The price distribution is right-skewed, with a smaller premium-priced segment.
- **Delivery time:** Most delivery times are around 40–60 minutes, with some restaurants taking more than 80 minutes.
- **Price and ratings:** Higher prices do not consistently correspond to higher ratings. Highly rated restaurants are also found at relatively affordable prices.
- **City comparison:** Kolkata has the most restaurant listings (1,346) and the highest average delivery time (67.81 minutes). Chennai has the highest average rating (3.78), while Mumbai has the highest average price (₹393.79). Surat has the lowest average price (₹270.17), and Ahmedabad has the shortest average delivery time (44.71 minutes).

### Business Implications

The analysis suggests that restaurants may benefit from competitive pricing, consistent food quality, and improved delivery efficiency. City-level differences also indicate that restaurant strategies should consider local market conditions rather than applying a single approach everywhere.

### Limitations and Future Scope

This analysis is based on the available dataset and describes associations rather than proving cause and effect. Further work could investigate cuisine-level rating differences, the relationship between customer review volume and ratings, and factors associated with longer delivery times.
